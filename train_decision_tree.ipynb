```python
from custom.util.data_manipulation import load_and_process_data, balance_dataset
from custom.models.ensemble import fit_ensemble_model, classify_then_regress
import json

_, _, numeric_features, categorical_features, df_train = load_and_process_data("data/claims_train.csv", True, "standard", True)
X_train, y_train, df_train = balance_dataset(df_train, numeric_features, categorical_features)
```

```python
X_test, y_test, numeric_features, categorical_features, df_test = load_and_process_data("data/claims_test.csv", True, "standard", True)
```

```python
from custom.models.decision_tree import DecisionTree

decision_tree = DecisionTree(min_to_split=2, min_samples_leaf=5, min_impurity_decrease=0.0, max_leaves=50)
decision_tree.fit(X_train, y_train)

```

```python
from custom.models.feedforward_neural_network import FeedForwardNeuralNetwork
import numpy as np

neural_network = FeedForwardNeuralNetwork(
    epochs = 100,
    learning_rate = np.float64(0.022216091081589597),
    optimizer = 'adam',
    regularization_setting = (2, np.float64(0.0008512374533940289)), 
    sizes_of_hidden_layers = [300, 200, 100, 50])
neural_network.fit(X_train, y_train)
```

```python
y_pred = decision_tree.predict(X_test)
y_pred2 = neural_network.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error

print("Test MSE:", mean_squared_error(y_test, y_pred))

print("Test MSE:", mean_squared_error(y_test, y_pred2))

```
